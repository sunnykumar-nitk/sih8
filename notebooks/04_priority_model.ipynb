# 04 -- Priority Scoring

Combines severity + population impact + infrastructure importance +
accessibility + disaster conditions + more into a single 0-100
priority score, using the transparent weighted formula in
`backend/recommendation/scoring.py`.

```python
import sys
sys.path.append("../backend")

from recommendation.scoring import calculate_priority, explain_score
import config

print("Current weights:", config.PRIORITY_WEIGHTS)
```

## Score the bridge example (2 ft crack, low severity, but high importance)

```python
bridge_site = {
    "damage_severity": 3,
    "population_impact": 7,
    "infrastructure_importance": 9,
    "accessibility": 3,
    "disaster_conditions": 4,
    "cascading_impact": 6,
}

result = calculate_priority(bridge_site)
print(explain_score(result))
```

## Score a batch of sites and rank them

```python
import pandas as pd

sites = [
    {"site_id": "BRIDGE_A", "damage_severity": 3, "population_impact": 7, "infrastructure_importance": 9, "accessibility": 3, "cascading_impact": 6},
    {"site_id": "BUILDING_B", "damage_severity": 9, "population_impact": 9, "infrastructure_importance": 6, "accessibility": 6},
    {"site_id": "ROAD_C", "damage_severity": 2, "population_impact": 3, "infrastructure_importance": 3, "accessibility": 2},
]

rows = []
for s in sites:
    r = calculate_priority(s)
    rows.append({"site_id": s["site_id"], "priority_score": r["priority_score"], "priority_level": r["priority_level"]})

pd.DataFrame(rows).sort_values("priority_score", ascending=False)
```
