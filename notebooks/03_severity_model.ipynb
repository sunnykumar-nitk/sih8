# 03 -- Severity Scoring

Converts a list of raw detections into a single 0-10 severity score,
label, and AI confidence. This is a deterministic formula, not a
separate trained model -- see `backend/models/severity_model.py`.

```python
import sys
sys.path.append("../backend")

from models.severity_model import calculate_severity
from models.damage_detector import get_detector

detector = get_detector()
detections = detector.detect("nepal_flood_site1.jpg")
severity = calculate_severity(detections)

print("Detections:", detections)
print()
print("Severity result:", severity)
```

## Compare severity across a few scenarios

```python
import pandas as pd

scenarios = ["nepal_flood_site1.jpg", "ahmedabad_crash_building.jpg", "generic_building.jpg"]
rows = []
for s in scenarios:
    dets = detector.detect(s)
    sev = calculate_severity(dets)
    rows.append({"scenario": s, **sev})

pd.DataFrame(rows)
```
