# 05 -- Team Allocation

Matches limited response teams to prioritized sites using specialization
match + distance. See `backend/recommendation/team_allocation.py`.

```python
import sys
sys.path.append("../backend")

from recommendation.team_allocation import allocate_teams

sites = [
    {"site_id": "HOSPITAL_ROAD", "priority_score": 91, "location": {"lat": 12.91, "lon": 74.79}, "required_expertise": "structural"},
    {"site_id": "BRIDGE_A", "priority_score": 68, "location": {"lat": 12.92, "lon": 74.80}, "required_expertise": "general"},
    {"site_id": "SCHOOL_C", "priority_score": 40, "location": {"lat": 12.93, "lon": 74.81}, "required_expertise": "general"},
]

teams = [
    {"team_id": "T1", "specialization": "structural", "location": {"lat": 12.90, "lon": 74.78}, "available": True},
    {"team_id": "T2", "specialization": "general", "location": {"lat": 12.95, "lon": 74.85}, "available": True},
]

assignments = allocate_teams(sites, teams)
for a in assignments:
    print(a)
```

## What happens with more sites than teams (5 sites, 2 teams)

```python
extra_sites = sites + [
    {"site_id": "ROAD_D", "priority_score": 55, "location": {"lat": 12.94, "lon": 74.82}},
    {"site_id": "ROAD_E", "priority_score": 30, "location": {"lat": 12.96, "lon": 74.83}},
]

assignments2 = allocate_teams(extra_sites, teams)
for a in assignments2:
    print(a)
# Sites without an available team are marked "UNASSIGNED" -- never silently dropped.
```
