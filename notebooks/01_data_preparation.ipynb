# 01 -- Data Preparation

This notebook prepares the supporting datasets used by the rest of the
pipeline: infrastructure importance data and team registry data.

It reads `backend/data/infrastructure.csv` and `backend/data/teams.csv`
(already included in the project) and does basic validation/cleaning.
Use this notebook as the place to add real GIS/population data later.

```python
import pandas as pd
import os

DATA_DIR = "../backend/data"
infra_path = os.path.join(DATA_DIR, "infrastructure.csv")
teams_path = os.path.join(DATA_DIR, "teams.csv")

infra_df = pd.read_csv(infra_path)
teams_df = pd.read_csv(teams_path)

print("Infrastructure records:", len(infra_df))
print("Team records:", len(teams_df))
infra_df.head()
```

## Basic validation -- check for missing coordinates or importance values

```python
missing_coords = infra_df[infra_df[["lat", "lon"]].isnull().any(axis=1)]
missing_importance = infra_df[infra_df["importance"].isnull()]

print("Rows missing coordinates:", len(missing_coords))
print("Rows missing importance score:", len(missing_importance))

assert len(missing_coords) == 0, "Fix missing coordinates before using this data downstream"
print("Infrastructure data looks valid.")
```

## Teams overview

```python
teams_df["equipment"] = teams_df["equipment"].str.split(",")
teams_df
```
